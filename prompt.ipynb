### Required Assignment 5.1: Will the Customer Accept the Coupon?

**Context**

Imagine driving through town and a coupon is delivered to your cell phone for a restaurant near where you are driving. Would you accept that coupon and take a short detour to the restaurant? Would you accept the coupon but use it on a subsequent trip? Would you ignore the coupon entirely? What if the coupon was for a bar instead of a restaurant? What about a coffee house? Would you accept a bar coupon with a minor passenger in the car? What about if it was just you and your partner in the car? Would weather impact the rate of acceptance? What about the time of day?

Obviously, proximity to the business is a factor on whether the coupon is delivered to the driver or not, but what are the factors that determine whether a driver accepts the coupon once it is delivered to them? How would you determine whether a driver is likely to accept a coupon?

**Overview**

The goal of this project is to use what you know about visualizations and probability distributions to distinguish between customers who accepted a driving coupon versus those that did not.

**Data**

This data comes to us from the UCI Machine Learning repository and was collected via a survey on Amazon Mechanical Turk. The survey describes different driving scenarios including the destination, current time, weather, passenger, etc., and then ask the person whether he will accept the coupon if he is the driver. Answers that the user will drive there ‘right away’ or ‘later before the coupon expires’ are labeled as ‘Y = 1’ and answers ‘no, I do not want the coupon’ are labeled as ‘Y = 0’.  There are five different types of coupons -- less expensive restaurants (under \$20), coffee houses, carry out & take away, bar, and more expensive restaurants (\$20 - $50).

**Deliverables**

Your final product should be a brief report that highlights the differences between customers who did and did not accept the coupons.  To explore the data you will utilize your knowledge of plotting, statistical summaries, and visualization using Python. You will publish your findings in a public facing github repository as your first portfolio piece.

**Investigating the Bar Coupons**

Now, we will lead you through an exploration of just the bar related coupons.  

1. Create a new `DataFrame` that contains just the bar coupons.


```python
databar = datac[datac["coupon"] == "Bar"]
databar.head()
```

2. What proportion of bar coupons were accepted?


```python
# Calculate the proportion of accepted coupons
total_observations = len(databar)
accepted_coupons = databar['Y'].sum()
proportion_accepted = accepted_coupons / total_observations

# Display results
print(f"Total observations: {total_observations}")
print(f"Accepted coupons: {accepted_coupons}")
print(f"Proportion of accepted coupons: {proportion_accepted:.4f} ({proportion_accepted*100:.2f}%)")
```

3. Compare the acceptance rate between those who went to a bar 3 or fewer times a month to those who went more.


```python
# Group A: ≤ 3 times/month
group_a = databar[databar['Bar'].isin(['never', 'less1', '1~3'])]
group_a_acceptance = group_a['Y'].sum() / len(group_a)

# Group B: > 3 times/month
group_b = databar[databar['Bar'].isin(['4~8', 'gt8'])]
group_b_acceptance = group_b['Y'].sum() / len(group_b)

# Display results
print(f"Group A (≤ 3 visits/month) acceptance rate: {group_a_acceptance:.2%}")
print(f"Group B (> 3 visits/month) acceptance rate: {group_b_acceptance:.2%}")
```

4. Compare the acceptance rate between drivers who go to a bar more than once a month and are over the age of 25 to the all others.  Is there a difference?


```python
# Group C: Goes to bar more than once/month AND over age 25
group_c = databar[
    databar['Bar'].isin(['1~3', '4~8', 'gt8']) &
    databar['age'].isin(['26', '31', '36', '41', '46', '50plus'])
]
group_c_acceptance = group_c['Y'].sum() / len(group_c)

# Group D: All others
group_d = databar[~databar.index.isin(group_c.index)]
group_d_acceptance = group_d['Y'].sum() / len(group_d)

# Print results
print(f"Group C (Bar > once/month & Age > 25) acceptance rate: {group_c_acceptance:.2%}")
print(f"Group D (All others) acceptance rate: {group_d_acceptance:.2%}")
```

5. Use the same process to compare the acceptance rate between drivers who go to bars more than once a month and had passengers that were not a kid and had occupations other than farming, fishing, or forestry.


```python
# Group E: More than once/month, no kids, not in farming/fishing/forestry
group_e = databar[
    databar['Bar'].isin(['1~3', '4~8', 'gt8']) &
    (databar['passanger'] != 'Kid(s)') &
    (databar['occupation'] != 'Farming Fishing & Forestry')
]
group_e_acceptance = group_e['Y'].sum() / len(group_e)

# Group F: All other drivers
group_f = databar[~databar.index.isin(group_e.index)]
group_f_acceptance = group_f['Y'].sum() / len(group_f)

# Print results
print(f"Group E (Bar > once/month, no kids, non-farming job): {group_e_acceptance:.2%}")
print(f"Group F (All others): {group_f_acceptance:.2%}")
```

6. Compare the acceptance rates between those drivers who:

- go to bars more than once a month, had passengers that were not a kid, and were not widowed *OR*
- go to bars more than once a month and are under the age of 30 *OR*
- go to cheap restaurants more than 4 times a month and income is less than 50K.



```python
# Group G: Goes to bars > once/month, no kids, not widowed
group_g = databar[
    databar['Bar'].isin(['1~3', '4~8', 'gt8']) &
    (databar['passanger'] != 'Kid(s)') &
    (databar['maritalStatus'] != 'Widowed')
]
group_g_acceptance = group_g['Y'].sum() / len(group_g)

# Group H: Goes to bars > once/month AND is under 30
group_h = databar[
    databar['Bar'].isin(['1~3', '4~8', 'gt8']) &
    databar['age'].isin(['below21', '21', '26'])
]
group_h_acceptance = group_h['Y'].sum() / len(group_h)

# Group J: Cheap restaurants > 4x/month AND income < $50K
group_j = databar[
    databar['RestaurantLessThan20'].isin(['4~8', 'gt8']) &
    databar['income'].isin([
        'Less than $12500',
        '$12500 - $24999',
        '$25000 - $37499',
        '$37500 - $49999'
    ])
]
group_j_acceptance = group_j['Y'].sum() / len(group_j)

# Print results with descriptive labels
print("Group G: Drivers who go to bars more than once a month, had passengers that were not a kid, and were not widowed")
print(f"→ Acceptance rate: {group_g_acceptance:.2%}\n")

print("Group H: Drivers who go to bars more than once a month and are under the age of 30")
print(f"→ Acceptance rate: {group_h_acceptance:.2%}\n")

print("Group J: Drivers who go to cheap restaurants more than 4 times a month and have income less than $50K")
print(f"→ Acceptance rate: {group_j_acceptance:.2%}")
```

### Independent Investigation

Using the bar coupon example as motivation, you are to explore one of the other coupon groups and try to determine the characteristics of passengers who accept the coupons.  

```python
# Group bar coupon data by passenger type
acceptance_by_passenger = databar.groupby('passanger')['Y'].mean().sort_values()
acceptance_counts = databar['passanger'].value_counts().reindex(acceptance_by_passenger.index)

# Create a bar plot
ax = acceptance_by_passenger.plot(kind='bar', color='skyblue', figsize=(10, 6))
plt.title('Bar Coupon Acceptance Rate by Passenger Type')
plt.xlabel('Passenger Type')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=45, ha='right')

# Add the count of each passenger type as text on the bars
for i, (rate, count) in enumerate(zip(acceptance_by_passenger, acceptance_counts)):
    ax.text(i, rate / 2, f"n={count}", ha='center', va='center', color='black', fontweight='bold')

plt.tight_layout()
plt.savefig("images/BarAcceptancePassengerType.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Define logical age order
age_order = ['below21', '21', '26', '31', '36', '41', '46', '50plus']

# Reindex acceptance rate and count using defined order
acceptance_by_age = databar.groupby('age')['Y'].mean().reindex(age_order)
acceptance_counts = databar['age'].value_counts().reindex(age_order)

# Plot
ax = acceptance_by_age.plot(kind='bar', color='skyblue', figsize=(10, 6))
plt.title('Bar Coupon Acceptance Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=45, ha='right')

# Add count labels
for i, (rate, count) in enumerate(zip(acceptance_by_age, acceptance_counts)):
    ax.text(i, rate / 2 if pd.notna(rate) else 0, f"n={count}", ha='center', va='center', color='black', fontweight='bold')

plt.tight_layout()
plt.savefig("images/BarAcceptanceAgeGroup.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Group bar coupon data by occupation
acceptance_by_occupation = databar.groupby('occupation')['Y'].mean().sort_values()
acceptance_counts = databar['occupation'].value_counts().reindex(acceptance_by_occupation.index)

# Create a bar plot
ax = acceptance_by_occupation.plot(kind='bar', color='skyblue', figsize=(12, 6))
plt.title('Bar Coupon Acceptance Rate by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=45, ha='right')

# Add the count of each occupation as text on the bars
for i, (rate, count) in enumerate(zip(acceptance_by_occupation, acceptance_counts)):
    ax.text(i, rate / 2, f"n={count}", ha='center', va='center', color='black', fontweight='bold')

plt.tight_layout()
plt.savefig("images/BarAcceptanceOccupation.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Define occupation-to-group mapping
occupation_groups = {
    'Management': 'Office Work',
    'Architecture & Engineering': 'Office Work',
    'Education': 'Office Work',
    'Healthcare Support': 'Office Work',
    'Legal': 'Office Work',
    'Business & Financial': 'Office Work',

    'Construction & Extraction': 'Skilled Trade',
    'Installation & Repair': 'Skilled Trade',
    'Production': 'Skilled Trade',
    'Transportation & Material Moving': 'Skilled Trade',

    'Building & Grounds Cleaning & Maintenance': 'Service',
    'Food Preparation & Serving Related': 'Service',
    'Personal Care & Service': 'Service',

    'Farming Fishing & Forestry': 'Agriculture',

    'Sales & Related': 'Other',
    'Office & Administrative Support': 'Other',
    'Protective Service': 'Other',

    'Student': 'Student',
    'Unemployed': 'Unemployed',
    'Retired': 'Unemployed'
}

# Apply mapping with .loc to avoid warnings
databar.loc[:, 'occupation_group'] = databar['occupation'].map(occupation_groups).fillna('Other')

# Set the display order
group_order = ['Office Work', 'Skilled Trade', 'Service', 'Agriculture', 'Other', 'Student', 'Unemployed']

# Calculate acceptance rates and sample sizes
acceptance_by_occ_group = databar.groupby('occupation_group')['Y'].mean().reindex(group_order)
acceptance_counts = databar['occupation_group'].value_counts().reindex(group_order)

# Create bar plot
ax = acceptance_by_occ_group.plot(kind='bar', color='skyblue', figsize=(10, 6))
plt.title('Bar Coupon Acceptance Rate by Occupation Group')
plt.xlabel('Occupation Group')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=45, ha='right')

# Add count labels
for i, (rate, count) in enumerate(zip(acceptance_by_occ_group, acceptance_counts)):
    ax.text(i, rate / 2 if pd.notna(rate) else 0, f"n={count}", ha='center', va='center', color='black', fontweight='bold')

plt.tight_layout()
plt.savefig("images/BarAcceptanceOccupationGroup.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Define the correct income order
income_order = [
    'Less than $12500',
    '$12500 - $24999',
    '$25000 - $37499',
    '$37500 - $49999',
    '$50000 - $62499',
    '$62500 - $74999',
    '$75000 - $87499',
    '$87500 - $99999',
    '$100000 or More'
]

# Reindex acceptance rate and count using defined order
acceptance_by_income = databar.groupby('income')['Y'].mean().reindex(income_order)
acceptance_counts = databar['income'].value_counts().reindex(income_order)

# Plot
ax = acceptance_by_income.plot(kind='bar', color='skyblue', figsize=(12, 6))
plt.title('Bar Coupon Acceptance Rate by Income')
plt.xlabel('Income')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=45, ha='right')

# Add count labels
for i, (rate, count) in enumerate(zip(acceptance_by_income, acceptance_counts)):
    ax.text(i, rate / 2 if pd.notna(rate) else 0, f"n={count}", ha='center', va='center', color='black', fontweight='bold')

plt.tight_layout()
plt.savefig("images/BarAcceptanceIncome.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Group by age and passenger, then calculate acceptance rate
acceptance_by_age_passenger = databar.groupby(['age', 'passanger'])['Y'].mean().unstack()

# Reorder age groups for logical display
age_order = ['below21', '21', '26', '31', '36', '41', '46', '50plus']
acceptance_by_age_passenger = acceptance_by_age_passenger.reindex(age_order)

# Create grouped bar chart
acceptance_by_age_passenger.plot(kind='bar', figsize=(12, 6), colormap='Blues')
plt.title('Bar Coupon Acceptance Rate by Age and Passenger Type')
plt.xlabel('Age Group')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Passenger Type')
plt.tight_layout()
plt.savefig("images/BarAcceptanceAgeAndPassengerType.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Columns to group by
group_cols = [
    'destination', 'passanger', 'weather', 'temperature', 'time',
    'expiration', 'gender', 'age', 'maritalStatus'
]

# Group and calculate acceptance rate and count
segment_stats_full = databar.groupby(group_cols)['Y'].agg(['mean', 'count']).reset_index()
segment_stats_full.columns = group_cols + ['acceptance_rate', 'sample_size']

# Filter to meaningful groups
segment_stats_full = segment_stats_full[segment_stats_full['sample_size'] >= 10]

# Get top 10 segments
top_segments_full = segment_stats_full.sort_values(by='acceptance_rate', ascending=False).head(10)

# Create readable labels for each row
top_segments_full['label'] = top_segments_full.apply(
    lambda row: f"{row['destination']}, {row['passanger']}, {row['weather']}, "
                f"{row['temperature']}°, {row['time']}, {row['expiration']} exp, "
                f"{row['gender']}, Age {row['age']}, {row['maritalStatus']}",
    axis=1
)

# Plot the top segments
plt.figure(figsize=(12, 9))
bars = plt.barh(top_segments_full['label'], top_segments_full['acceptance_rate'], color='skyblue')
plt.xlabel('Acceptance Rate')
plt.title('Top 10 Bar Coupon Acceptance Segments by Context & Demographics')
plt.gca().invert_yaxis()

# Annotate with sample size
for bar, size in zip(bars, top_segments_full['sample_size']):
    plt.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height() / 2,
             f"n={size}", va='center', fontsize=10)

plt.tight_layout()
plt.savefig("images/BarAcceptanceTop10.png", dpi=300, bbox_inches='tight')
plt.show()
```
